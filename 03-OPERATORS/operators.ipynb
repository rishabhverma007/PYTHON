```python
a = int(input("Enter a number a:"))
b = int(input("Enter a number b:"))
add= a+b
sub = a-b
multi = a*b
division = a/b
floor_division = a//b
modulus = a%b
print("ADDITION: ",add)
print("SUBSTRACTION: ",sub)
print("MULTIPLICATION: ",multi)
print("DIVISION: ",division)
print("FLOOR DIVISION: ", floor_division)
print("MODULUS: ",modulus)
```
